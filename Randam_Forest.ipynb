```python
# ==========================================
# RANDOM FOREST CLASSIFICATION
# ==========================================

# Step 1: Import Libraries

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score


# Step 2: Create Dataset

data = {
    'Patient_ID': ['P1', 'P2', 'P3', 'P4', 'P5',
                   'P6', 'P7', 'P8', 'P9', 'P10',
                   'P11', 'P12'],

    'Age': [40, 35, 55, 60, 75, 45, 50, 65, 30, 70, 58, 42],

    'Weight': [70, 60, 68, 79, 90, 65, 72, 85, 55, 88, 78, 62],

    'BP': [120, 145, 130, 140, 135, 125, 150, 145, 110, 160, 138, 118],

    'HeartDisease': ['No', 'Yes', 'No', 'Yes', 'Yes', 'No',
                     'Yes', 'Yes', 'No', 'Yes', 'Yes', 'No']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)


# Step 3: Separate Features and Target

X = df[['Age', 'Weight', 'BP']]

y = df['HeartDisease']


# Step 4: Split Dataset into Training and Testing

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

print("\nTraining Data:")
print(X_train)

print("\nTesting Data:")
print(X_test)


# Step 5: Create Random Forest Model

model = RandomForestClassifier(
    n_estimators=5,
    max_features='sqrt',
    bootstrap=True,
    random_state=42
)


# Step 6: Train the Model

model.fit(X_train, y_train)

print("\nRandom Forest Model Trained Successfully!")


# Step 7: Make Predictions

y_pred = model.predict(X_test)

print("\nActual Values:")
print(list(y_test))

print("\nPredicted Values:")
print(y_pred)


# Step 8: Calculate Accuracy

accuracy = accuracy_score(y_test, y_pred)

print("\nModel Accuracy:")
print(accuracy * 100, "%")


# Step 9: Display Individual Decision Trees

print("\nNumber of Decision Trees:",
      len(model.estimators_))


# Step 10: Predict New Patient

new_patient = pd.DataFrame({
    'Age': [55],
    'Weight': [80],
    'BP': [145]
})

prediction = model.predict(new_patient)

print("\nNew Patient Details:")
print(new_patient)

print("\nFinal Heart Disease Prediction:")
print(prediction[0])


# Step 11: Display Feature Importance

importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': model.feature_importances_
})

print("\nFeature Importance:")
print(importance.sort_values(
    by='Importance',
    ascending=False
))


# ==========================================
# END OF RANDOM FOREST PRACTICAL
# ==========================================
```
